```python
import os
import pytesseract
from PIL import Image, ImageOps

folder_paths = [
    '/content/drive/MyDrive/Project2_Data/whatsapp_messages',
    '/content/drive/MyDrive/Project2_Data/twitterX roman urdu tweets'
]

data = []

for folder in folder_paths:
    if os.path.exists(folder):
        for filename in os.listdir(folder):
            if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                file_path = os.path.join(folder, filename)
                try:
                    # Open the image and convert it to grayscale
                    img = Image.open(file_path).convert('L')

                    # INVERT COLORS: This forces a dark theme screenshot to look like a standard white page
                    # (If your screenshots are already white, just delete or comment out the line below!)
                    img = ImageOps.invert(img)

                    # Extract the text
                    extracted_text = pytesseract.image_to_string(img).strip()

                    # Evaluate each line Tesseract found separately
                    lines = extracted_text.split('\n')

                    for line in lines:
                        clean_sentence = " ".join(line.split())
                        words = clean_sentence.split()

                        # FILTER: Only keep it if it's a real sentence (4 or more words)
                        # This skips UI noise like "Lee", "11:45 PM", or "Retweet"
                        if len(words) >= 4:
                            entry = {
                                'sentence': clean_sentence,
                                'words': words,
                                'labels': [] # Ready for your tags
                            }
                            data.append(entry)
                except Exception as e:
                    print(f"Could not read {filename}: {e}")

print(f"Successfully filtered images and found {len(data)} potential sentences!")
if len(data) > 0:
    print("\nHere is your first filtered entry:")
    print(data[0])
```

```python
import pandas as pd

# This creates rows directly from your scraped 'data' list
rows = []
for entry in data:
    for word in entry['words']:
        rows.append({
            'sentence': entry['sentence'],
            'word': word,
            'label': '' # Left blank for you to fill!
        })

# Convert to a DataFrame and save as a CSV template
df_template = pd.DataFrame(rows)
df_template.to_csv('dataset_template.csv', index=False, encoding='utf-8')

print(f"Template created successfully with {len(df_template)} rows!")
print("Download 'dataset_template.csv' from your left file sidebar, fill in the labels, and upload it back!")
```
